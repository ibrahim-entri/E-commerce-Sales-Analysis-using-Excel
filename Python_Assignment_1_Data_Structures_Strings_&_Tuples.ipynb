```python
# String 1
str1 = "Hello "
str1
```

```python
# Take name from user
name = input("Enter your Name: ")
name
```

```python
# Concatenate
result = str1 + name
result
```

```python
# Add second string
str3 = ", welcome to Python programming"

final_result = result + str3
final_result
```

```python
text = final_result
text
```

```python
# a. Print the first character of the string.
"first character:", text[0]
```

```python
# b. Last character
"last character:",text[-1]
```

```python
# c. First 5 characters
"first 5 characters:", text[:5]
```

```python
# d. Last 11 characters
"last 11 characters:", text[-11:]
```

```python
# e. Reverse string
"reverse string:",text[::-1]
```

```python
# f. Extract word "Python"
"Word Python:", text[ text.find("Python") : text.find("Python") + 6 ]
```

```python
# String Methods:
strM = "Python beginner tutorial"
strM
```

```python
# a. Uppercase
"Uppercase:", strM.upper()
```

```python
# b. Lowercase
"lowercase:", strM.lower()
```

```python
# c. Capitalize
"Capitalize:", strM.capitalize()
```

```python
# d. Count 't'
"count of t:", strM.count("t")
```

```python
# e. Replace Python
"replace Python", strM.replace("Python","Machine Learning")
```
